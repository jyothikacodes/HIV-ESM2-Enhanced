## Rare Mutation Frequency Preparation

This enhancement replaces the separate rare-mutation notebook and writes reusable outputs under `data/rare_mutations/`.


```python
# Rare mutation frequency preparation
# Produces frequency matrices, rarity weights, rare masks, and summary tables for the attention model.
RUN_RARE_MUTATION_PREP = False

if RUN_RARE_MUTATION_PREP:
    import pickle
    import numpy as np
    import pandas as pd
    from src.data_processing import load_unified_data
    from src.feature_engineering import HIV_PROTEASE_REFERENCE, HIV_RT_REFERENCE
    from src.rare_mutations import (
        compute_mutation_frequencies,
        compute_rare_mutation_weights,
        compute_rare_mutation_mask,
        get_rare_mutation_summary,
        normalize_weights_for_attention,
    )

    unified_data = load_unified_data(PROCESSED_DIR)
    references = {
        'PI': HIV_PROTEASE_REFERENCE,
        'NRTI': HIV_RT_REFERENCE,
        'NNRTI': HIV_RT_REFERENCE,
    }
    rare_dir = DATA_DIR / 'rare_mutations'
    rare_dir.mkdir(parents=True, exist_ok=True)

    for drug_class, data in unified_data.items():
        sequences = data['sequences']
        reference = references[drug_class]
        out_dir = rare_dir / drug_class
        out_dir.mkdir(parents=True, exist_ok=True)

        frequencies = compute_mutation_frequencies(sequences, reference)
        weights = compute_rare_mutation_weights(sequences, reference, frequencies=frequencies)
        masks = compute_rare_mutation_mask(sequences, reference, frequencies=frequencies)
        summary = get_rare_mutation_summary(sequences, reference, frequencies=frequencies)
        normalized_weights = [normalize_weights_for_attention(w, method='softmax') for w in weights]

        np.save(out_dir / 'frequencies.npy', frequencies)
        np.save(out_dir / 'weights.npy', np.array(weights, dtype=object), allow_pickle=True)
        np.save(out_dir / 'normalized_weights.npy', np.array(normalized_weights, dtype=object), allow_pickle=True)
        np.save(out_dir / 'masks.npy', np.array(masks, dtype=object), allow_pickle=True)
        summary.to_csv(out_dir / 'rare_mutation_summary.csv', index=False)
        with open(out_dir / 'rare_mutation_outputs.pkl', 'wb') as f:
            pickle.dump({
                'frequencies': frequencies,
                'weights': weights,
                'normalized_weights': normalized_weights,
                'masks': masks,
                'summary': summary,
            }, f)

        print(f'{drug_class}: saved rare mutation outputs to {out_dir}')
else:
    print('Set RUN_RARE_MUTATION_PREP = True to compute rare mutation frequency outputs.')
```
